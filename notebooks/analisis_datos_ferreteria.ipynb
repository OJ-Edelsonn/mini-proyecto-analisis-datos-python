## 11. Limpieza y transformacion de datos

En esta fase se preparan los datos para el analisis:

- normalizacion de nombres de columnas,
- limpieza de textos,
- conversion de fechas y valores numericos,
- validacion del calculo de ventas,
- integracion de ventas con el catalogo de productos,
- creacion de variables derivadas para analisis.


```python
import unicodedata

def normalizar_columna(nombre_columna):
    texto = unicodedata.normalize("NFKD", nombre_columna)
    texto = "".join(caracter for caracter in texto if not unicodedata.combining(caracter))
    texto = texto.strip().lower().replace(" ", "_")
    return texto

productos_limpio = productos.copy()
ventas_limpio = ventas.copy()

productos_limpio.columns = [normalizar_columna(col) for col in productos_limpio.columns]
ventas_limpio.columns = [normalizar_columna(col) for col in ventas_limpio.columns]

print(productos_limpio.columns.tolist())
print(ventas_limpio.columns.tolist())
```

### 11.1 Limpieza de textos y conversion de tipos

Despues de normalizar los nombres de columnas, se limpian espacios innecesarios en campos de texto y se convierten fechas y columnas numericas al tipo correcto.


```python
# Limpieza de espacios en columnas de texto
for columna in productos_limpio.select_dtypes(include="object").columns:
    productos_limpio[columna] = productos_limpio[columna].astype(str).str.strip()

for columna in ventas_limpio.select_dtypes(include="object").columns:
    ventas_limpio[columna] = ventas_limpio[columna].astype(str).str.strip()

# Conversion de tipos
productos_limpio["costo"] = pd.to_numeric(productos_limpio["costo"], errors="coerce")
productos_limpio["stock"] = pd.to_numeric(productos_limpio["stock"], errors="coerce").astype("Int64")

ventas_limpio["fecha"] = pd.to_datetime(ventas_limpio["fecha"], errors="coerce")
ventas_limpio["cantidad"] = pd.to_numeric(ventas_limpio["cantidad"], errors="coerce").astype("Int64")
ventas_limpio["precio_unitario"] = pd.to_numeric(ventas_limpio["precio_unitario"], errors="coerce")
ventas_limpio["total"] = pd.to_numeric(ventas_limpio["total"], errors="coerce")

print(productos_limpio.dtypes)
print()
print(ventas_limpio.dtypes)
```

### 11.2 Validacion del total de ventas

Se recalcula el total de cada linea de venta para verificar si coincide con el valor informado en el dataset original.


```python
ventas_limpio["total_calculado"] = (
    ventas_limpio["cantidad"].astype(float) * ventas_limpio["precio_unitario"]
).round(2)

ventas_limpio["diferencia_total"] = (
    ventas_limpio["total"] - ventas_limpio["total_calculado"]
).round(2)

errores_total = ventas_limpio[ventas_limpio["diferencia_total"].abs() > 0.01]

print("Registros con diferencia en total:", len(errores_total))
display(ventas_limpio[["cantidad", "precio_unitario", "total", "total_calculado", "diferencia_total"]].head())
```

### 11.3 Integracion de ventas con catalogo de productos

Se integra el archivo de ventas con el catalogo de productos para agregar informacion clave como ID de producto, costo unitario y stock actual.


```python
productos_base = productos_limpio.rename(columns={"nombre": "producto"})

ventas_enriquecidas = ventas_limpio.merge(
    productos_base[["id_producto", "producto", "categoria", "costo", "stock"]],
    on="producto",
    how="left",
    suffixes=("_venta", "_catalogo")
)

ventas_sin_producto = ventas_enriquecidas[ventas_enriquecidas["id_producto"].isna()]

print("Ventas sin producto encontrado en catalogo:", len(ventas_sin_producto))
display(ventas_enriquecidas.head())
```

### 11.4 Creacion de variables derivadas

Se crean nuevas variables que facilitan el analisis posterior: margen estimado, margen porcentual y columnas de tiempo para tendencias.


```python
ventas_enriquecidas["categoria"] = ventas_enriquecidas["categoria_catalogo"].fillna(
    ventas_enriquecidas["categoria_venta"]
)

ventas_enriquecidas["margen_unitario"] = (
    ventas_enriquecidas["precio_unitario"] - ventas_enriquecidas["costo"]
).round(2)

ventas_enriquecidas["margen_estimado"] = (
    ventas_enriquecidas["margen_unitario"] * ventas_enriquecidas["cantidad"].astype(float)
).round(2)

ventas_enriquecidas["margen_pct"] = (
    ventas_enriquecidas["margen_estimado"] / ventas_enriquecidas["total"]
).round(4)

ventas_enriquecidas["anio"] = ventas_enriquecidas["fecha"].dt.year
ventas_enriquecidas["mes"] = ventas_enriquecidas["fecha"].dt.month
ventas_enriquecidas["periodo"] = ventas_enriquecidas["fecha"].dt.to_period("M").astype(str)
ventas_enriquecidas["trimestre"] = ventas_enriquecidas["fecha"].dt.to_period("Q").astype(str)

display(
    ventas_enriquecidas[
        [
            "fecha",
            "periodo",
            "producto",
            "categoria",
            "cantidad",
            "precio_unitario",
            "costo",
            "total",
            "margen_unitario",
            "margen_estimado",
            "margen_pct",
        ]
    ].head()
)
```

### 11.5 Generacion del dataset limpio

Se seleccionan y ordenan las columnas finales para crear un dataset limpio, enriquecido y listo para analisis.


```python
ventas_limpias = ventas_enriquecidas[
    [
        "fecha",
        "periodo",
        "anio",
        "mes",
        "trimestre",
        "id_producto",
        "producto",
        "categoria",
        "cantidad",
        "precio_unitario",
        "total",
        "total_calculado",
        "diferencia_total",
        "costo",
        "margen_unitario",
        "margen_estimado",
        "margen_pct",
        "stock",
    ]
].sort_values(["fecha", "producto"]).reset_index(drop=True)

productos_limpios = productos_base.copy()
productos_limpios["valor_inventario"] = (
    productos_limpios["costo"] * productos_limpios["stock"]
).round(2)

ventas_limpias.to_csv(DATA_PROCESSED / "ventas_limpias.csv", index=False, encoding="utf-8-sig")
productos_limpios.to_csv(DATA_PROCESSED / "productos_limpios.csv", index=False, encoding="utf-8-sig")

print("Archivo generado:", DATA_PROCESSED / "ventas_limpias.csv")
print("Archivo generado:", DATA_PROCESSED / "productos_limpios.csv")
print("Ventas limpias:", ventas_limpias.shape)
print("Productos limpios:", productos_limpios.shape)
```

### 11.6 Reporte de calidad posterior a la limpieza

Se genera un resumen de validaciones clave para confirmar que el dataset limpio conserva integridad y consistencia.


```python
reporte_calidad = pd.DataFrame(
    [
        {
            "validacion": "productos_total",
            "valor": len(productos_limpios),
            "estado": "OK",
            "detalle": "Cantidad de productos en catalogo limpio",
        },
        {
            "validacion": "ventas_total",
            "valor": len(ventas_limpias),
            "estado": "OK",
            "detalle": "Cantidad de registros en ventas limpias",
        },
        {
            "validacion": "nulos_ventas_limpias",
            "valor": int(ventas_limpias.isna().sum().sum()),
            "estado": "OK" if int(ventas_limpias.isna().sum().sum()) == 0 else "Revisar",
            "detalle": "Valores nulos en dataset limpio de ventas",
        },
        {
            "validacion": "nulos_productos_limpios",
            "valor": int(productos_limpios.isna().sum().sum()),
            "estado": "OK" if int(productos_limpios.isna().sum().sum()) == 0 else "Revisar",
            "detalle": "Valores nulos en dataset limpio de productos",
        },
        {
            "validacion": "ventas_sin_producto",
            "valor": int(ventas_limpias["id_producto"].isna().sum()),
            "estado": "OK" if int(ventas_limpias["id_producto"].isna().sum()) == 0 else "Revisar",
            "detalle": "Ventas sin producto asociado",
        },
        {
            "validacion": "totales_mal_calculados",
            "valor": int((ventas_limpias["diferencia_total"].abs() > 0.01).sum()),
            "estado": "OK" if int((ventas_limpias["diferencia_total"].abs() > 0.01).sum()) == 0 else "Revisar",
            "detalle": "Diferencia entre total original y total recalculado",
        },
    ]
)

display(reporte_calidad)

reporte_calidad.to_csv(PROJECT_ROOT / "reports" / "quality_report.csv", index=False, encoding="utf-8-sig")
print("Reporte generado:", PROJECT_ROOT / "reports" / "quality_report.csv")
```

## 12. Analisis descriptivo

En esta fase se calculan metricas generales y rankings principales para entender el comportamiento comercial de la ferreteria despues de la limpieza de datos.


```python
from pathlib import Path

import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

ventas_limpias = pd.read_csv(DATA_PROCESSED / "ventas_limpias.csv")
productos_limpios = pd.read_csv(DATA_PROCESSED / "productos_limpios.csv")

ventas_limpias["fecha"] = pd.to_datetime(ventas_limpias["fecha"])

print("Ventas limpias:", ventas_limpias.shape)
print("Productos limpios:", productos_limpios.shape)
```

### 12.1 KPIs generales

Se calculan indicadores principales de ventas, unidades, transacciones, ticket promedio y margen estimado.


```python
ventas_totales = ventas_limpias["total"].sum()
unidades_vendidas = ventas_limpias["cantidad"].sum()
transacciones = len(ventas_limpias)
promedio_venta = ventas_limpias["total"].mean()
precio_promedio_ponderado = ventas_totales / unidades_vendidas
margen_estimado_total = ventas_limpias["margen_estimado"].sum()
margen_estimado_pct = margen_estimado_total / ventas_totales

kpis_generales = pd.DataFrame(
    {
        "KPI": [
            "Ventas totales",
            "Unidades vendidas",
            "Transacciones",
            "Promedio de venta por linea",
            "Precio promedio ponderado",
            "Margen estimado",
            "Margen estimado %",
        ],
        "Valor": [
            ventas_totales,
            unidades_vendidas,
            transacciones,
            promedio_venta,
            precio_promedio_ponderado,
            margen_estimado_total,
            margen_estimado_pct,
        ],
    }
)

kpis_generales["Valor_formateado"] = [
    f"S/ {ventas_totales:,.2f}",
    f"{unidades_vendidas:,.0f}",
    f"{transacciones:,.0f}",
    f"S/ {promedio_venta:,.2f}",
    f"S/ {precio_promedio_ponderado:,.2f}",
    f"S/ {margen_estimado_total:,.2f}",
    f"{margen_estimado_pct:.2%}",
]

kpis_generales
```

**Lectura inicial:** El dataset registra S/ 128,182.96 en ventas, 8,056 unidades vendidas y 1,800 lineas de venta. El margen estimado total es de S/ 35,775.12, equivalente a 27.91% de las ventas. Esta lectura confirma que los resultados calculados en Python son consistentes con los KPIs del Proyecto 1 en Power BI.


### 12.2 Productos mas vendidos

Se identifican los productos con mayor venta total y mayor cantidad de unidades vendidas. Esto permite diferenciar productos que generan mas ingresos de productos que tienen mayor rotacion.


```python
ranking_productos = (
    ventas_limpias
    .groupby(["id_producto", "producto", "categoria"], as_index=False)
    .agg(
        ventas_totales=("total", "sum"),
        unidades_vendidas=("cantidad", "sum"),
        transacciones=("producto", "count"),
        margen_estimado=("margen_estimado", "sum"),
        precio_promedio=("precio_unitario", "mean"),
    )
)

ranking_productos["margen_pct"] = (
    ranking_productos["margen_estimado"] / ranking_productos["ventas_totales"]
).round(4)

ranking_productos = ranking_productos.sort_values("ventas_totales", ascending=False)

ranking_productos.head(10)
```

```python
ranking_productos_unidades = ranking_productos.sort_values(
    "unidades_vendidas", ascending=False
)

ranking_productos_unidades.head(10)
```

**Lectura:** El ranking por ventas muestra que Cemento bolsa 42.5 kg es el principal producto del negocio por ingresos. Sin embargo, al revisar unidades vendidas aparecen productos de menor precio con alta rotacion, como Arena fina saco 40 kg y Ladrillo pandereta unidad. Esta diferencia es importante porque un producto puede liderar en ingresos sin necesariamente ser el de mayor rotacion operativa.


### 12.3 Categorias con mayor venta

Se analiza el desempeno por categoria para identificar que lineas de negocio concentran mayores ventas, unidades y margen estimado.


```python
resumen_categorias = (
    ventas_limpias
    .groupby("categoria", as_index=False)
    .agg(
        ventas_totales=("total", "sum"),
        unidades_vendidas=("cantidad", "sum"),
        transacciones=("producto", "count"),
        margen_estimado=("margen_estimado", "sum"),
        precio_promedio=("precio_unitario", "mean"),
    )
)

resumen_categorias["participacion_ventas_pct"] = (
    resumen_categorias["ventas_totales"] / resumen_categorias["ventas_totales"].sum()
).round(4)

resumen_categorias["margen_pct"] = (
    resumen_categorias["margen_estimado"] / resumen_categorias["ventas_totales"]
).round(4)

resumen_categorias = resumen_categorias.sort_values(
    "ventas_totales", ascending=False
)

resumen_categorias
```

**Lectura:** Materiales de construccion es la categoria con mayor venta total y concentra una parte importante del negocio. Herramientas electricas aparece como una categoria de alto ingreso pese a vender pocas unidades, lo que sugiere productos de mayor ticket promedio. En cambio, categorias como Electricidad o Tornilleria pueden tener mayor movimiento operativo por volumen de unidades.


### 12.4 Analisis mensual de ventas

Se agrupan las ventas por periodo mensual para observar evolucion en ventas, unidades, transacciones y margen estimado.


```python
resumen_mensual = (
    ventas_limpias
    .groupby("periodo", as_index=False)
    .agg(
        ventas_totales=("total", "sum"),
        unidades_vendidas=("cantidad", "sum"),
        transacciones=("producto", "count"),
        margen_estimado=("margen_estimado", "sum"),
        ticket_promedio=("total", "mean"),
    )
)

resumen_mensual["variacion_ventas_pct"] = (
    resumen_mensual["ventas_totales"].pct_change()
).round(4)

resumen_mensual["media_movil_3m"] = (
    resumen_mensual["ventas_totales"].rolling(window=3, min_periods=1).mean()
).round(2)

resumen_mensual
```

```python
mejor_mes = resumen_mensual.loc[resumen_mensual["ventas_totales"].idxmax()]
peor_mes = resumen_mensual.loc[resumen_mensual["ventas_totales"].idxmin()]

print("Mejor mes:")
display(mejor_mes.to_frame().T)

print("Peor mes:")
display(peor_mes.to_frame().T)
```

**Lectura:** El resumen mensual permite identificar meses de mayor y menor desempeno comercial. Esta informacion sera util para construir visualizaciones de tendencia y para comparar si los picos de ventas corresponden a mayor cantidad de unidades, mayor ticket promedio o una combinacion de ambos factores.


### 12.5 Productos con mayor margen estimado

Ademas de analizar ventas, se revisan los productos que mas aportan margen estimado. Esto ayuda a diferenciar productos que venden mucho de productos que contribuyen mas a la rentabilidad.


```python
ranking_margen = ranking_productos.sort_values(
    "margen_estimado", ascending=False
)

ranking_margen.head(10)
```

```python
comparativo_ventas_margen = ranking_productos[
    [
        "producto",
        "categoria",
        "ventas_totales",
        "unidades_vendidas",
        "margen_estimado",
        "margen_pct",
    ]
].head(10)

comparativo_ventas_margen
```

**Lectura:** Los productos lideres por ventas tambien tienden a concentrar margen estimado, pero no siempre en la misma proporcion. Este analisis permite priorizar productos no solo por ingresos, sino tambien por contribucion economica. Para una ferreteria, esta mirada es clave al decidir abastecimiento, promociones o acciones comerciales.
