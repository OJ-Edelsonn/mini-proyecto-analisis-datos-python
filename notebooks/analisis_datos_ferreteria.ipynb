## 2. Importacion de librerias


```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
FIGURES = PROJECT_ROOT / "reports" / "figures"

print("Ruta del proyecto:", PROJECT_ROOT)
```

## 3. Carga de datos


```python
productos = pd.read_csv(DATA_RAW / "productos.csv")
ventas = pd.read_csv(DATA_RAW / "ventas.csv")

print("Productos:", productos.shape)
print("Ventas:", ventas.shape)
```

## 8. Revision de duplicados


```python
duplicados = pd.DataFrame({
    "revision": [
        "Productos duplicados por id_producto",
        "Productos duplicados por nombre",
        "Filas duplicadas en ventas",
    ],
    "valor": [
        productos["id_producto"].duplicated().sum(),
        productos["nombre"].duplicated().sum(),
        ventas.duplicated().sum(),
    ],
})

duplicados
```
